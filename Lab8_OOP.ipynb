```python
class Shape:
    def area(self):
        return "Calculating area..."


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


# Create objects
circle = Circle(5)
rectangle = Rectangle(4, 6)

# Output
print(f"Circle area: {circle.area()}")
print(f"Rectangle area: {rectangle.area()}")
```

```python
class Device:
    def power_on(self):
        return "Device is powering on..."


class Smartphone(Device):
    def power_on(self):
        return "Smartphone is booting up 📱"


class Laptop(Device):
    def power_on(self):
        return "Laptop is starting up 💻"


# Function that demonstrates polymorphism
def start_device(device):
    print(device.power_on())


# Create objects
phone = Smartphone()
laptop = Laptop()

# Pass different objects to same function
start_device(phone)
start_device(laptop)
```

```python
class TemperatureConverter:
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (celsius * 9/5) + 32


celsius = 25  # fixed value = instant run

fahrenheit = TemperatureConverter.celsius_to_fahrenheit(celsius)

print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # addition (+)
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    # subtraction (-)
    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def display(self):
        print(f"({self.x}, {self.y})")


# create vectors
v1 = Vector(5, 3)
v2 = Vector(2, 7)

# operations
v_add = v1 + v2
v_sub = v1 - v2

# output
print("Vector 1:")
v1.display()

print("Vector 2:")
v2.display()

print("Addition:")
v_add.display()

print("Subtraction:")
v_sub.display()
```

```python
from abc import ABC, abstractmethod


class Appliance(ABC):
    @abstractmethod
    def turn_on(self):
        pass


class Refrigerator(Appliance):
    def turn_on(self):
        return "Refrigerator is now cooling."


# create object
fridge = Refrigerator()

# call method
print(fridge.turn_on())
```

```python
class A:
    def show(self):
        return "Class A"


class B(A):
    def show(self):
        return "Class B"


class C(A):
    def show(self):
        return "Class C"


class D(B, C):
    pass


# create object
obj = D()

# method resolution result
print(obj.show())

# show MRO
print(D.mro())
```

```python
class ATM:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient balance"
        self.balance -= amount
        return f"Withdrawn {amount}, Remaining balance: {self.balance}"


atm = ATM(1000)
print(atm.withdraw(200))
```
